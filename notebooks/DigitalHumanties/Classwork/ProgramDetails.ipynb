Create a chart showing who puts in the most effort. Effort can be find from number of hours.(For example Student A has the huighest study hours meaning they put in the most effort.)

```python
# can use to work with tabular data
import pandas as pd 
df = pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.title("How many hours students studied")
plt.ylabel("Hours_studied")
plt.xlabel("Student")
avg_student= df.groupby("Student")["Hours_Studied"].mean()
avg_student.plot(kind="bar")
plt.show()
plt.show()
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
#colors= plt.cm.plasma(range(len(avg_hours)))
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#FFC0CB")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
#s=df["Years_of_practice"]*60 this means if student has more years of practice their bubble is bigger
#alpha=0.6 makes the bubbles bit transparent
plt.title("Hours studied vs grades")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, x="Hours_Studied", y="Grade")
plt.title("Study patterns and perfomances")
plt.show()
```

```python
plt.Figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot for hours studied and grades")
plt.show
```

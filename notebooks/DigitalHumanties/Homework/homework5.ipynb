I used my own dataset from previous homework. with pd.read_csv i made my csv file with dataset visible here 


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("dataset.csv")
df
```

I made a bar chart that includes only birth years and major work counts to see who had most works and when were they born

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Birth Year", y="Major Works Count", legend=False)
plt.title("Poets birth years and their major works count")
plt.xlabel("Birth Year")
plt.ylabel("Major Works Count")
plt.show()
```

I can see that the oldest poet is born in 1770 and most major works has poet that was born in 1799 having more than 80 works.

Then i made pie chart

```python
df.plot(kind="pie", y="Birth Year", labels=df["Major Works Count"], legend=False, autopct="%1.1f%%")
plt.title("Poets Birth years and their major works count")
```

I decided to make next chart that includes genres and major works count to see which genre was used most/less and how many works wrote poets with these genres.

```python
plt.scatter(df["Genre"], df["Major Works Count"], s=df["Major Works Count"]*60, alpha=0.2)
#s=df["Major Works Count"]*60 this means if poet has more major works their bubble is bigger
#alpha=0,6 makes the bubbles bit transparent
plt.title("Genre vs Major works count")
plt.xlabel("Genre")
plt.ylabel("Major works count")
```

I can see that poets mostly used poetry and novel genres althought its clear that novels were used more because it has more major works count. Essay genre was used less.


I made box chart including only birth years

```python
import seaborn as sns
sns.boxplot(data=df, y="Birth Year")
plt.title("Distribution of birth years")
plt.show()
```

Then i made bar chart again but this time with names and major works to see who exactly wrote more. I added purple color to my chart


```python
import matplotlib.pyplot as plt
avg_work= df.groupby("Name")["Major Works Count"].mean()
#colors= plt.cm.plasma(range(len(avg_work)))
avg_work.plot(kind="bar", x="Name", y="Major_Works_Count", color="#800080")
```

By this chart i can be sure that Honore de Balzac made more major works than others.
